**PRAKTIKUM**

```python
# Praktikum 01 – Sequential Search pada Data Angka

def sequential_search(data, target):
  for i in range(len(data)):
    if data[i] == target:
      return f"Data {target} ditemukan di indeks {i}"
  return f"Data {target} tidak ditemukan"

angka = [10, 20, 30, 40, 50]
print("Data:", angka)
print(sequential_search(angka, 30))
print(sequential_search(angka, 100))
```

```python
# Praktikum 02 – Binary Search pada Data Angka

def binary_search(data, target):
  left, right = 0, len(data) -1
  while left<=right:
    mid = (left + right) //2
    if data[mid] == target:
      return f"Data {target} ditemukan di indeks {mid}"
    elif data[mid] < target:
      left = mid + 1
    else:
      right = mid - 1
  return f"Data {target} tidak ditemukan"

angka = [10, 20, 30, 40, 50]
print("Data:", angka)
print(binary_search(angka, 40))
print(binary_search(angka, 100))
```

```python
# Praktikum 03 – Sequential Search untuk Data Nama

def sequential_search_nama(data, target):
  for i in range(len(data)):
    if data[i].lower() == target.lower():
      return f"Nama {target} ditemukan di indeks {i}"
  return f"Nama {target} tidak ditemukan"

nama = ["Andi", "Budi", "Cici", "Doni"]
print("Data:", nama)
print(sequential_search_nama(nama, "Cici"))
print(sequential_search_nama(nama, "Eka"))
```

```python
# Praktikum 04 – Studi Kasus Nilai Mahasiswa dengan Binary Search


def binary_search_nilai(data, target):
  left, right = 0, len(data)-1
  while left <= right:
    mid = (left + right) // 2
    if data[mid] == target:
      return f"Nilai {target} ditemukan di indeks {mid}"
    elif data[mid] < target:
      left = mid + 1
    else:
      right = mid - 1
  return f"Nilai {target} tidak ditemukan"

nilai = [60, 70, 75, 80, 85, 90]
print("Data nilai:", nilai)
print(binary_search_nilai(nilai, 85))
print(binary_search_nilai(nilai, 65))
```

```python
# Praktikum 05

import time

def sequential_search_count(data, target):
    count = 0
    for i in range(len(data)):
        count += 1
        if data[i] == target:
            return count
    return count

def binary_search_count(data, target):
    left = 0
    right = len(data) - 1
    count = 0

    while left <= right:
        count += 1
        mid = (left + right) // 2

        if data[mid] == target:
            return count
        elif data[mid] < target:
            left = mid + 1
        else:
            right = mid - 1

    return count

data = list(range(1, 100001))
target = 99999

print("Perbandingan Sequential vs Binary Search")

start = time.time()
seq_count = sequential_search_count(data, target)
seq_time = time.time() - start

start = time.time()
bin_count = binary_search_count(data, target)
bin_time = time.time() - start

print(f"\nTarget dicari : {target}")
print("\n--- Sequential Search ---")
print("Langkah :", seq_count)
print("Waktu   :", seq_time, "detik")

print("\n--- Binary Search ---")
print("Langkah :", bin_count)
print("Waktu   :", bin_time, "detik")
```

```python
# Praktikum 06 – Kombinasi Sorting dan Searching

import random

def insertion_sort(data):
  for i in range(1, len(data)):
    key = data[i]
    j = i - 1
    while j >= 0 and data[j] > key:
      data[j + 1] = data[j]
      j -= 1
    data[j + 1] = key
  return data

def binary_search(data, target):
  left, right = 0, len(data)-1
  while left <= right:
    mid = (left + right) // 2
    if data[mid] == target:
      return f"{target} ditemukan di indeks {mid}"
    elif data[mid] < target:
      left = mid + 1
    else:
      right = mid - 1
  return f"{target} tidak ditemukan"

data = [random.randint(1, 100) for _ in range(15)]
print("Data acak sebelum diurutkan:", data)
urut = insertion_sort(data)
print("Data setelah diurutkan:", urut)
target = 56
print(binary_search(urut, target))
```

```python
# Penugasan 01 - Variasi Sequential Search

def sequential_search_count(data, target):
    count = 0
    for i in range(len(data)):
        count += 1
        if data[i] == target:
            return i, count
    return -1, count

data = [10, 20, 30, 40, 50]
target = 50

idx, langkah = sequential_search_count(data, target)
print("Hasil Sequential Search:", idx)
print("Jumlah perbandingan:", langkah)
```

```python
# Penugasan 02 - Variasi Binary Search

def binary_search_count(data, target):
    left, right = 0, len(data) - 1
    langkah = 0

    while left <= right:
        langkah += 1
        mid = (left + right) // 2

        if data[mid] == target:
            return mid, langkah
        elif data[mid] < target:
            left = mid + 1
        else:
            right = mid - 1

    return -1, langkah

data = [10, 20, 30, 40, 50]
target = 10

idx, langkah = binary_search_count(data, target)
print("Hasil Binary Search:", idx)
print("Jumlah iterasi:", langkah)
```

```python
# Penugasan 03 - Studi Kasus Nama Mahasiswa

def sequential_search_nama(data, target):
    for i in range(len(data)):
        if data[i].lower() == target.lower():
            return i
    return -1

n = int(input("Berapa banyak nama mahasiswa? "))

mahasiswa = []
for i in range(n):
    nama = input(f"Nama ke-{i+1}: ")
    mahasiswa.append(nama)

cari = input("Masukkan nama yang ingin dicari: ")

idx = sequential_search_nama(mahasiswa, cari)

if idx != -1:
    print(f"Nama '{cari}' ditemukan di indeks {idx}")
else:
    print(f"Nama '{cari}' TIDAK ditemukan")
```

```python
# Penugasan 04 - Studi Kasus Barang Toko

def binary_search_barang(barang, harga):
    left, right = 0, len(barang) - 1

    while left <= right:
        mid = (left + right) // 2
        if barang[mid][1] == harga:
            return mid
        elif barang[mid][1] < harga:
            left = mid + 1
        else:
            right = mid - 1
    return -1

barang = [
    ("Pensil", 2000),
    ("Buku", 5000),
    ("Penghapus", 3000),
    ("Pulpen", 4000),
]

barang.sort(key=lambda x: x[1])

target_harga = 4000
idx = binary_search_barang(barang, target_harga)

print("Data barang urut:", barang)
if idx != -1:
    print("Barang ditemukan:", barang[idx])
else:
    print("Barang tidak ditemukan")
```

```python
# Penugasan 05 - Analisis Kompleksitas – Perbandingan Waktu (100, 1000, 10000)

import time

def sequential_search(data, target):
    for i in range(len(data)):
        if data[i] == target:
            return i
    return -1

def binary_search(data, target):
    left, right = 0, len(data)-1
    while left <= right:
        mid = (left+right)//2
        if data[mid] == target:
            return mid
        elif data[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1

sizes = [100, 1000, 10000]

print("Ukuran | Waktu Sequential | Waktu Binary")
print("-----------------------------------------")

for size in sizes:
    data = list(range(1, size+1))
    target = size

    # Sequential
    t1 = time.time()
    sequential_search(data, target)
    t2 = time.time()

    # Binary
    t3 = time.time()
    binary_search(data, target)
    t4 = time.time()

    print(f"{size:<6} | {t2-t1:.6f} detik     | {t4-t3:.6f} detik")
```
