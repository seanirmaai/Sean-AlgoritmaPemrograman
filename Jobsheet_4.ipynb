**PRAKTIKUM**

**Perulangan For**

```python
for i in range(1, 6):
    print(i)
```

**Perulangan While**

```python
n = int(input("Masukkan angka: "))
while n > 0:
  print(n)
  n -= 1
```

**Nested Loop**

```python
for i in range(1, 6):
    for j in range(i):
        print("*", end="")
    print()
```

**Break dan Continue**

```python
#Kode phyton break

for i in range(1, 11):
    if i == 5:
        print("Angka 5 ditemukan")
        break
```

```python
#Kode phyton continue

for i in range(1, 6):
    if i % 2 == 1:
        continue
    print(i)
```

**For pada Dictionary**

```python
data = {"Nama": "Andi", "Umur": 20, "Prodi": "TRK"}

for k, v in data.items():
  print (k, ":", v)
```

**For dengan Enumerate**

```python
daftar = ["Andi", "Budi", "Cici"]

for i, nama in enumerate(daftar, start=1):
  print(i, nama)
```

**Studi kasus: Program kasir multi-item**

```python
total = 0

while True:
  barang =  input("Masukkan nama barang (ketik 'Selesai' untuk berhenti): ")
  if barang.lower() == "selesai":
    break
  harga = float(input(f"Masukkan harga {barang}"))
  jumlah = int(input(f"Masukkan jumlah {barang}"))
  subtotal = harga * jumlah
  total += subtotal

print("\n===== Struk Belanja =====")
print(f"Total Belanja : Rp {total}")
```

```python
#Hitung faktorial

n = int(input("Masukkan bilangan: "))
hasil = 1

while n > 0:
  hasil *= n
  n -= 1

print(f"Hasil faktorial:", hasil)
```

```python
#Pola segitiga terbalik

for i in range(5, 0, -1):
  for j in range(i):
    print("*", end="")
  print()
```

```python
#Cari Bilangan Prima dengan Break

bilangan = int(input("Masukkan bilangan: "))

if bilangan > 1:
    for i in range(2, bilangan):
      if bilangan % i == 0:
            print(f"{bilangan} bukan bilangan prima (habis dibagi {i})")
            break  # hentikan loop lebih awal

    else:
           print(bilangan, "adalah bilangan prima.")
else:
    print(bilangan, "bukan bilangan prima (kurang dari 2).")
```

```python
#Tampilkan Hanya Bilangan Genap dengan Continue

for i in range(1, 21):
    if i % 2 != 0:
        continue  # Lewati bilangan ganjil
    print(i)
```

```python
#Iterasi Dictionary Mahasiswa

mahasiswa = {"Nama": "Budi", "Umur": 21, "Prodi": "Teknik Komputer"}

for key, value in mahasiswa.items():
    print(f"{key} : {value}")
```

```python
#Enumerate Daftar Barang

barang = ["Buku", "Pulpen", "Tas"]

for i, nama in enumerate(barang, start=1):
    print(i, nama)
```

```python
# Program kasir sederhana

total = 0
belanja = []

while True:
    barang = input("Masukkan nama barang (ketik 'selesai' untuk berhenti): ")
    if barang.lower() == "selesai":
        break
    harga = float(input(f"Masukkan harga {barang}: "))
    jumlah = int(input(f"Masukkan jumlah {barang}: "))
    subtotal = harga * jumlah
    belanja.append((barang, harga, jumlah, subtotal))
    total += subtotal

# Hitung diskon
diskon = 0
if total >= 200000:
    diskon = total * 0.10

total_akhir = total - diskon

# Tampilkan struk belanja
print("\n===== STRUK BELANJA =====")
for b in belanja:
    print(f"{b[0]} x{b[2]} = Rp {b[3]:,.0f}")
print("---------------------------")
print(f"Total Belanja : Rp {total:,.0f}")
print(f"Diskon (10%)  : Rp {diskon:,.0f}")
print(f"Total Bayar   : Rp {total_akhir:,.0f}")
print("===========================")
```
