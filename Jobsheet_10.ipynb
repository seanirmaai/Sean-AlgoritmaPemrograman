```python
# Praktikum 01

def bubble_sort(data):
  n = len(data)
  for i in range(n):
    for j in range (0, n - i - 1):
      if data [j] > data[j + 1]:
        data[j], data[j + 1] = data[j + 1], data[j]
  return data

angka = [64, 34, 25, 12, 22, 11, 90]
print("Data sebelum diurutkan:", angka)
print("Data setelah Bubble Sort:", bubble_sort(angka))
```

```python
# Praktikum 02

def selection_sort(data):
    n = len(data)
    for i in range(n):
        min_index = i
        for j in range(i + 1, n):
            if data[j] < data[min_index]:
                min_index = j
        data[i], data[min_index] = data[min_index], data[i]
    return data

angka = [64, 25, 12, 22, 11]
print("Data sebelum diurutkan:", angka)
print("Data setelah Selection Sort:", selection_sort(angka))
```

```python
# Praktikum 03

def insertion_sort(data):
  n = len(data)
  for i in range(1, n):
    key = data[i]
    j = i - 1
    while j >= 0 and data[j] > key:
      data[j + 1] = data[j]
      j -= 1
    data[j + 1] = key
  return data

angka = [12, 11, 13, 5, 6]
print("Data sebelum diurutkan:", angka)
print("Data setelah Insertion Sort:", insertion_sort(angka))
```

```python
# Praktikum 04

def bubble_sort_nilai(data):
  n = len(data)
  for i in range(n):
    for j in range(0, n - i -1):
      if data[j][1]< data[j + 1][1]:
        data[j], data[j + 1] = data[j + 1], data[j]
  return data

mahasiswa = [("Andi", 80), ("Budi", 70), ("Cici", 90), ("Doni", 85)]
print("Data sebelum diurutkan:", mahasiswa)
print("Data setelah diurutkan (desc):", bubble_sort_nilai(mahasiswa))
```

```python
# Praktikum 06

def bubble_sort_key(data, key_index=0, reverse=False):
  n = len(data)
  for i in range(n):
    for j in range(0, n - i - 1):
      if reverse:
        if data[j][key_index] < data[j + 1][key_index]:
          data[j], data[j + 1] = data[j + 1], data[j]
        else:
          if data[j][key_index] > data[j + 1][key_index]:
            data[j], data[j + 1] = data[j + 1], data[j]
  return data

barang = [("Buku", 15000), ("Pensil", 5000), ("Tas", 75000), ("Pulpen", 10000)]
print("Urut berdasarkan nama (asc):", bubble_sort_key(barang.copy(), 0))
print("Urut berdasarkan harga (desc):", bubble_sort_key(barang.copy(), 1, reverse=True))
```

```python
# Penugasan 01 - Eksperimen Bubble Sort

def bubble_sort(data):
    n = len(data)
    for i in range(n):
        for j in range(0, n - i - 1):
            # Ubah tanda perbandingan
            if data[j] < data[j + 1]:
                data[j], data[j + 1] = data[j + 1], data[j]
    return data

angka = [76, 27, 16, 18, 98, 66]
print("Data sebelum diurutkan:", angka)
print("Data setelah Bubble Sort (Descending):", bubble_sort(angka))
```

```python
# Penugasan 02 - Eksperimen Selection Sort

def selection_sort(data):
    n = len(data)
    for i in range(n - 1, 0, -1):
        max_index = 0
        for j in range(1, i + 1):
            if data[j] > data[max_index]:
                max_index = j
        data[i], data[max_index] = data[max_index], data[i]
    return data

angka = [97, 77, 12, 10, 21]
print("Data sebelum diurutkan:", angka)
print("Data setelah Selection Sort (Max di akhir):", selection_sort(angka))
```

```python
# Penugasan 03 - Eksperimen Insertion Sort

def insertion_sort(data):
    n = len(data)
    for i in range(1, n):
        key = data[i]
        j = i - 1

        while j >= 0 and data[j].lower() > key.lower():
            data[j + 1] = data[j]
            j -= 1

        data[j + 1] = key

    return data

nama_mahasiswa = input("Masukkan daftar nama mahasiswa (pisahkan dengan koma): ").split(",")

nama_mahasiswa = [nama.strip() for nama in nama_mahasiswa]

print("Data sebelum diurutkan:", nama_mahasiswa)
print("Data setelah Insertion Sort (A-Z):", insertion_sort(nama_mahasiswa))
```

```python
#Penugasan 5 - Analisis Kompleksitas

import time, random

def bubble_sort(data):
  n = len(data)
  for i in range(n):
    for j in range(0, n - i - 1):
      if data[j] > data[j + 1]:
        data[j], data[j + 1] = data[j + 1], data[j]
  return data

def selection_sort(data):
  n = len(data)
  for i in range(n):
    min_index = i
    for j in range (i + 1, n):
      if data[j] < data[min_index]:
        min_index = j
    data[i], data[min_index] = data[min_index], data[i]
  return data

def insertion_sort(data):
  n = len(data)
  for i in range (1, n):
    key = data[i]
    j = i - 1
    while j >= 0 and data[j] > key:
      data[j + 1] = data[j]
      j -= 1
    data[j + 1] = key
  return data

def analisis_kompleks(ukuran):
    for i in ukuran:
      data = [random.randint(1,1000) for i in range(i)]
      print(f"\n=== Ukuran data: {i} ===")
      print(data)
      for func in [bubble_sort, selection_sort, insertion_sort]:
          data = data.copy()
          start = time.time()
          end = time.time()
          print(f"{func.__name__} membutuhkan waktu {end - start:.10f} detik")

ukuran = [10, 50, 100]
analisis_kompleks(ukuran)
```
