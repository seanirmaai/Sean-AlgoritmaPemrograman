**PRAKTIKUM**

```python
#Praktikum 01 - Fungsi Dasar dengan 'def'

def sapa():
  print("Halo, selamat datang di Politeknik Negeri Semarang")

sapa()
```

```python
#Praktikum 02 - Fungsi dengan Parameter dan Return

def tambah(a, b):
  return a + b

hasil = tambah (7, 3)
print ("Hasil penjumlahan:", hasil)
```

```python
#Praktikum 03 - Fungsi dengan Default Argument

def sapa(nama="Mahasiswa"):
  print(f"Halo, {nama}!")

sapa()        #tanpa argumen
sapa("Andi")  #dengan argumen
```

```python
#Praktikum 04 - Fungsi dengan *args dan **kwargs

def jumlahkan(*args):
  return sum(args)

def biodata(**kwargs):
  for key, value in kwargs.items():
    print(f"{key}: {value}")

print("Total:", jumlahkan(1, 2, 3, 4, 5))
biodata(Nama="Budi", Umur=20, Prodi="TRK")
```

```python
#Praktikum 05 - Scope Variabel

x =10

def ubah():
  x = 5
  print("Nilai lokal:", x)

ubah()
print("Nilai global:", x)
```

```python
#Praktikum 06 - Fungsi Lambda

kali = lambda x, y: x * y
print("Hasil perkalian:", kali(4, 5))
```

```python
#Praktikum 07 - Fungsi Decorator

def decorator_sapa(fungsi):
  def wrapper():
    print("Sebelum fungsi dijalankan...")
    fungsi()
    print("Setelah fungsi dijalankan...")
  return wrapper

@decorator_sapa
def sapa():
  print("Halo, Mahasiswa Polines!")

sapa()
```

```python
#Praktikum 08 - Studi Kasus: Menghitung rata-rata nilai mahasiswa

def hitung_rata(data):
  return sum(data) / len(data)

nilai = [80, 90, 85]
rata = hitung_rata(nilai)

print("Rata-rata nilai:", rata)
```

**PENUGASAN**

```python
#Penugasan 01 - Fungsi Sederhana

def perkenalan():
  print("Halo, nama saya Sean, saya dari prodi Teknologi Rekayasa Komputer")

def perkenalan2():
  print("Kamu ridho ngga aku makan?")

def perkenalan3():
  print("Sorry sorry sorry")

perkenalan()
perkenalan2()
perkenalan3()
```

```python
#Penugasan 02 - Fungsi dengan Parameter dan Return

def luas_persegi(panjang, lebar):
  return panjang * lebar

print ("Luas 1:", luas_persegi(5, 3))
print ("Luas 2:", luas_persegi(7, 3))
print ("Luas 3:", luas_persegi(6, 3))
```

```python
#Penugasan 03 - Fungsi dengan Default Argument

def salam(waktu="Pagi", nama="Mahasiswa"):
  print(f"Selamat {waktu}, {nama}!")

salam()
salam("siang")
salam("malam", "Dori")
```

```python
#Penugasan 04 - Fungsi dengan args

def jumlahkan_semua(*angka):
  return sum(angka)

print("Jumlah 3 Bilangan:", jumlahkan_semua(1, 2, 3))
print("Jumlah 5 Bilangan:", jumlahkan_semua(4, 5, 6, 7, 8))
print("Jumlah 7 Bilangan:", jumlahkan_semua(10, 20, 30, 40, 50, 60, 70))
```

```python
#Penugasan 05 - Fungsi dengan kwargs

def info_barang(**data):
  for key, value in data.items():
    print(f"{key}: {value}")

info_barang(Nama="Buku", Harga=15000, Stok="30")
```

```python
#Penugasan 06 - Scope Variable

x = 100

def ubah():
  x = 50
  print("Nilai lokal:", x)

ubah()
print("Nilai global:", x)
```

```python
#Penugasan 07 - Fungsi Lambda

kuadrat = lambda x: x*x
print("Hasil kuadrat dari 5 adalah:", kuadrat(5))

keliling_lingkaran = lambda r: 22/7*(2*r)
print("Keliling lingkaran dengan jari jari 7 adalah:", keliling_lingkaran(7))
```

```python
#Penugasan 08 - Fungsi Decorator

def logger(fungsi):
  def uhuy():
    print("Memulai fungsi...")
    fungsi()
    print("Fungsi selesai dijalankan...")
  return uhuy

@logger
def halo():
  print("Halo, teman teman semuanya")

halo()
```

```python
#Penugasan 09 - Studi Kasus Nyata – Program Nilai

def hitung_rata(data):
    return sum(data)  / len(data)

def status_kelulusan(rata):
    if rata >= 60:
        return "Lulus"
    else:
        return "Tidak Lulus"

input_tugas = float(input("Masukkan nilai Tugas (0-100): "))
input_ujian = float(input("Masukkan nilai Ujian (0-100): "))

nilai = [90, 88, 95]
rata = hitung_rata(nilai)
hitung_nilai_akhir = lambda nilai_tugas, nilai_ujian: (0.4 * nilai_tugas) + (0.6 * nilai_ujian)
nilai_akhir = hitung_nilai_akhir (input_tugas, input_ujian)

print("Hasil rata - rata",rata)
print("Status Kelulusan:", status_kelulusan(rata))
print(f"Nilai Akhir (40% Tugas + 60% Ujian): {nilai_akhir:.2f}")
```
